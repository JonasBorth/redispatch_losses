```python
from open_mastr import Mastr
from sqlalchemy import create_engine
from sqlalchemy import URL

url_object = URL.create(
    "postgresql+psycopg2",
    username="postgres",
    password="postgres",
    host="localhost",
    database="redispatch_losses",
)

engine = create_engine(
    url_object,
    pool_recycle=3600,
    pool_size=10,
    max_overflow=20,
    #    pool_timeout=,
    echo=True,
)
db = Mastr(engine=engine)
db.download(data=["wind","solar"])
```
